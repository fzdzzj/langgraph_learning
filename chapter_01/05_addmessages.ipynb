```python
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage
from langgraph.graph import add_messages

left = [
    SystemMessage(content="你是一个专业的翻译",id = '1'),
    HumanMessage(content="翻译：你好",id = '2'),
    AIMessage(content="你好",id = '3'),
]
right = [
    HumanMessage(content='我是老王，你是小王',id = '2'),
    AIMessage(content="好的，我记住了",id = '3'),
    HumanMessage(content="你是谁",id = '4'),
    AIMessage(content="我是老王",id = '5'),
]
merged = add_messages(left,right)
print(merged)
```
